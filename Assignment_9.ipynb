```python
import random  # Generate random number between 1 and 10
secret_number = random.randint(1, 10)
attempts = 3

while attempts > 0:
    quess_number = int(input("Guess a number between 1 & 10: "))

    if quess_number < 1 or quess_number > 10:  # if guess is out of the range
        print("Your input is out of the range. The range is between 1&10")
        continue

    if quess_number == secret_number:  # compare guess with secret number
        print("Congrats, you guessed the exact number")
        break

    elif quess_number > secret_number:  # if guess value is too high
        print("Too hight. Try again")

    else:
        print("Too low. Try again")  # if guess value is too low

    attempts -= 1  # reduce attempts

else:
    print("Better luck next time!")  # if the user runs out of attempts
```

```python

def calculate_bmi(weight, height_cm): # Define function
    height_m = height_cm / 100   # convert cm to meters
    return weight / (height_m ** 2)

weight = float(input("Enter your weight in KG: ")) # Prompt user for input
height_cm = float(input("Enter your height in CM: "))


bmi = calculate_bmi(weight, height_cm) # Calculate BMI
print("Your BMI is:", round(bmi, 2))
```

Multiplication Table Generator (For Loop)

```python
# Ask the user for a number
num = int(input("Enter the number for which you want the multiplication table: "))

# Use a for loop to generate the table
for i in range(1, 11):
    # Print using simple string concatenation
    print(str(num) + " x " + str(i) + " = " + str(num * i))
```
