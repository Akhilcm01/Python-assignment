List creation

```python
#Create a list named age_list with five integer elements. For eg., [24, 25, 26, 27, 28]

age_list=[24, 25, 26, 27, 28] #code to create a list of age
print(age_list)
```

```python
#Create a list named name_list with five string elements.

name_list=["Rahul","Sanjay","Rohan","Riya","Zera"] #code to create a list of names
print(name_list)
```

```python
#Append the string "Yazhini" to name_list.

name_list.append("Yazhini") #code to appending existing list
print(name_list)
```

```python
#Insert the element 30 at index 2 in age_list.

age_list.insert(2,30) #code to isert an element in to the specific position of the list by specifying index number
print(age_list)
```

```python
#Remove the string "Yazhini" from name_list.

name_list.remove("Yazhini") #remove is removing an element from list using specifying the element
print(name_list)
```

```python
#Pop the last element from age_list.

age_list.pop(5)
print(age_list) #pop is removing an element from list using index position
```

```python
#Extend the age_list with additional ages [29, 30, 26].

age_list.extend([29, 30, 26]) #code to extend an existing list to the end. List is mutable data structure
print(age_list)
```

```python
#Sort age_list in descending order.

age_list.sort(reverse=True) #code to display a list in largest to smallest order
print(age_list)
```

```python
#Find Max age, Min age and sum of all ages from age_list.

max_age=max(age_list)
print("Maximum age is",max_age) #code to print maximum age from the list
```

```python
min_age=min(age_list)
print("Minimum age is",min_age) #code to print minimum age from the list
```

```python
sum_age=sum(age_list)
print("Summ of all ages is",sum_age)
```

```python
#Print the first element of name_list.
print(name_list[0]) #to print the first element of the list using index position
```

```python
#Print the last element of name_list.

print(name_list[-1]) #code to print last element in the list using reverse indexing
```

```python
#Create two sets:

set1={1, 3, 5, 7, 9} #code to create set
set2={2, 3, 5, 8, 10} #code to create set

print(set1)

print(set2)
```

```python
#Compute and print the union and intersection of set1 and set2.


union_set = set1.union(set2) # Union: elements present in either set1 or set2

intersection_set = set1.intersection(set2) # Intersection: elements present in both set1 and set2

print("Union of set1 and set2:", union_set)
print("Intersection of set1 and set2:", intersection_set)
```
