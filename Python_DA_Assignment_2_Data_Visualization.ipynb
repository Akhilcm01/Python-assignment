```python
#fill na in payment

#find number of na values

df["payment"].isnull().sum()

#Fill na in payment with key word unknowm

df["payment"] = df["payment"].fillna("unknown")

#check if nulls replaced

df["payment"].isnull().sum()

```

```python
#fill na in pickup_zone

#find number of na values

print(df["pickup_zone"].isnull().sum())

#imputation using text "Unkown pick up zone"

df["pickup_zone"]=df["pickup_zone"].fillna("Unknown Pickup zone")

#check if imputation is successful

print(df["pickup_zone"].isnull().sum())
```

```python
#fillna in dropoff_zone

#find number of na values

df["dropoff_zone"].isnull().sum()

#imputation using text "Unkown dropoff zone"

df["dropoff_zone"]=df["dropoff_zone"].fillna("Unkown dropoff zone")

#check if imputation is successful

df["dropoff_zone"].isna().sum()
```

```python
#fillna in pickup_borough column


#find number of na values

df["pickup_borough"].isnull().sum()

#imputation using text "pickup_borough not available"

df["pickup_borough"]=df["pickup_borough"].fillna("pickup_borough not available")

#check if imputation is successful

df["pickup_borough"].isnull().sum()
```

```python
#fillna in dropoff_borough column

#find number of na values

df["dropoff_borough"].isnull().sum()

#imputation using text "dropoff_borough not available"

df["dropoff_borough"]=df["dropoff_borough"].fillna("dropoff_borough not available")

#check if imputation is successful

df["dropoff_borough"].isnull().sum()
```

```python
#Recheck info

df.info()
```

```python
#Matplotlib line plot

#convert pick up column to time format

df["pickup"]=pd.to_datetime(df["pickup"])

#sort chronologically

df=df.sort_values("pickup")

df.dtypes

plt.plot(df["pickup"],df["fare"],linestyle="--",linewidth=.5)
plt.title("Fare vs Time")
plt.xlabel("Time")
plt.ylabel("Fare")
plt.show()
```

```python
#Bar chart

#farebyborough

farebyborough=df.groupby("pickup_borough")["fare"].mean()
categories = farebyborough.index
values = farebyborough.values
plt.bar(categories, values,color='r')
plt.title("Fare by Borough")
plt.xlabel("Borough")
plt.ylabel("Fare")
plt.show()
```

```python
#pie plot

payment_counts = df["payment"].value_counts()

plt.pie(payment_counts.values, labels=payment_counts.index,autopct='%1.3f%%')
plt.title("Payment Types")
plt.show()
```

```python
#histogram

plt.hist(df["distance"],bins=30,color="yellow",edgecolor="black")
plt.title("Distribution of Trip Distance")
plt.xlabel("Distance (miles)")
plt.ylabel("Frequency")
plt.show()
```

```python
#boxplot - Use pickup_borough as the categorical axis and tip as the numeric axis.

# Get the unique boroughs
boroughs = df["pickup_borough"].unique()

# Build a list of tip values for each borough
tips_by_borough = [df[df["pickup_borough"] == b]["tip"] for b in boroughs]

# Plot boxplot
plt.figure(figsize=(8,5))
plt.boxplot(tips_by_borough, labels=boroughs)
plt.title("Tip by Borough")
plt.xlabel("Borough")
plt.ylabel("Tip")
plt.show()
```

```python
#Visualizations using Seaborn:
"""Create a count plot to visualize the number of trips in each pickup_borough. The
x-axis should represent the boroughs, and the y-axis should show the count of trips.
"""

#Count Plot
sns.countplot(x="pickup_borough",data=df)
plt.title("Number of Trips by Borough")
plt.xlabel("Borough")
plt.ylabel("Count")
plt.show()





```

```python
#scatter plot
""" Plot a scatter plot to show the relationship between distance and fare. Use
distance on the x-axis and fare on the y-axis to visualize any correlation. Color the
points based on the pickup_borough to differentiate the trips by their respective
boroughs."""

sns.set(style="darkgrid")
sns.scatterplot(x="distance",y="fare",data=df,hue="pickup_borough")
plt.title("Distance vs Fare")
plt.show()
```

```python
#Violin Plot
"""Plot a violin plot to show the distribution of fare for each payment method. Use the
payment method as the categorical axis and fare as the numeric axis to visualize its
distribution"""

sns.violinplot(x="payment",y="fare",data=df)
plt.title("Fare by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Fare")
plt.show()
```

```python
#heatmap
""" Plot a heatmap to visualize the correlation between numerical variables such as
distance, fare, tip, tolls, and total. Use a correlation matrix to highlight the
relationships """

numeric_cols = ["distance", "fare", "tip", "tolls", "total"]
corr = df[numeric_cols].corr()
plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Taxi Variables")
plt.show()

```

```python
#Pair Plot

"""Create a pair plot to visualize the pairwise relationships between distance, fare, tip,
and total. Color the data points according to the pickup_zone method to compare
how different zones affect these variables"""

sns.pairplot(df, vars=["distance", "fare", "tip", "total"], hue="pickup_zone")
plt.show()


```
