String Concatenation

```python
string1="Hello"
string2=" Zara"
strconct=string1+string2 #concats or joins two strings using + function
print(strconct)

```

```python
string3=",welcome to Python programming"
print(strconct,string3) #concats or joins two strings using , function
```

String Slicing and Indexing

```python
slice1="Hello Zara, welcome to Python programming"

print(slice1[0])  #Print the first character of the string.
```

```python
print(slice1[-1])  #Print the last character of the string
```

```python
print(slice1[:5])  #Print the first 5 characters of the string.
```

```python
print(slice1[-12:])  #Print the last 11 characters of the string.
```

```python
print(slice1[::-1]) #Print the string in reverse using slice
```

```python
reverse1="".join(reversed(slice1)) #Print the string in reverse using reversed function
print(reverse1)
```

```python
print(slice1[23:30]) #Use slicing and print the word “Python” from the existing string.
```

String Methods

```python
strM ="Python beginner tutorial"
print(strM.upper())  #Convert the sentence to uppercase.
```

```python
print(strM.lower()) #Convert the sentence to lowercase.
```

```python
print(strM.capitalize()) #Use Capitalize and return the sentence to the original input form.
```

```python
print(strM.count('t')) #Count the total number of occurrences of character ‘t’ in the string.
```

```python
print(strM.replace("Python","Machine learning"))
#Replace all occurrences of “Python” with “Machine Learning” in the input string strM = “Python beginner tutorial”
```

Tuples (Creation, Modification and Access)

```python
t1=(10, 20, 30)
t2=(40, 50, 60)
t_combine=t1+t2
print(t_combine)

#Concatenate the two tuples and store it in “t_combine"
```

```python
print(t_combine*3)

#Repeat the elements of “t_combine” 3 times
```

```python
print(t_combine[2])

# Access the 3rd element from “t_combine”
```

```python
print(t_combine[:3])
#Access the first three elements from “t_combine”
```

```python
print(t_combine[-3:])

#Access the last three elements from “t_combine”
```
