```python
"""
Problem Statement: To learn how to use NumPy for creating arrays,
performing basic array operations, and conducting numerical computations
through analyzing temperature data.
Scenario: You are given a list of daily average temperatures (in degrees Celsius)
recorded over two weeks. Your task is to use NumPy to analyze this data.
"""
```

```python
import numpy as np

#Create a 1D NumPy array named temperatures_w1 for Week 1 with the following values:


temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
print(temperatures_w1)
```

```python
#Inspect the shape, data type, and the number of elements in the array.
#Inspect the shape of the temperatures_w1 array.

print(temperatures_w1.shape) #row,col
print(temperatures_w1.dtype)
print(temperatures_w1.size)
```

```python
#Convert the temperatures_w1 from Celsius to Fahrenheit using the
#formula: Fahrenheit = (Celsius * 9/5) + 32

temperatures_f=(temperatures_w1*9/5)+32
print(temperatures_f)

#Find the maximum, minimum, and mean temperatures for the week.

max_temp=temperatures_f.max()
min_temp=temperatures_f.min()
mean_temp=temperatures_f.mean()
print("Maximum temprature is: ",max_temp)
print("Minumim temperature is :",min_temp)
print("Mean of temperatureis:",mean_temp)




```

```python
import pandas as pd

#Create a Pandas Series named marks with the following data:

marks=pd.Series([95, 92, 89, 85, 80],index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
print(marks)
```

```python
#Use the integer index position to access the mark of the 1st rank student.

print(marks[0])
```

```python
#Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels

top3_marks=marks.loc[['Rank1', 'Rank2', 'Rank3']]
print(top3_marks)
```

```python
#Use the iloc accessor to retrieve the mark of the 3rd rank student

third_mark=marks.iloc[2]
print(third_mark)

```

```python
#Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90

high_marks=marks[marks>90]
print(high_marks)
```

```python
#Modify the mark of the student with the 1st rank to 100.


marks.loc['Rank1']=100
print(marks)
```

```python
#Remove the entry corresponding to the last rank from the Series.

marks=marks.drop('Rank5')
print(marks)
```

```python
#Compute the CGPA by dividing each mark in the Series by 10.
marks=marks/10
print(marks)
```

```python
transactions=pd.DataFrame({'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109,110],'ProductCategory':['Electronics', 'Clothing', 'Electronics', 'Furniture',
'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],'Region':['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West','South','North'],'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]})

print(transactions)
```

```python
#Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

print(transactions.head())
print(transactions.tail())
print(transactions.shape)
print(transactions.columns)
print(transactions.dtypes)
```

```python
#Display only the 'ProductCategory' and 'Amount' columns

print(transactions[['ProductCategory','Amount']])
```

```python
#Retrieve the last 3 columns of the DataFrame using iloc or loc.

print(transactions.iloc[:,-3:])
```

```python
#Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

north_transactions=transactions[(transactions['Region']=='North') & (transactions['Amount']>200)]
print(north_transactions)
```

```python
#Find the value counts for the 'ProductCategory' column

product_counts=transactions['ProductCategory'].value_counts()
print(product_counts)
```

```python
#Find the unique values in the 'Region' column

unique_regions=transactions['Region'].unique()
print(unique_regions)
```

```python
#Group by 'Region' and find the mean amount for each region.

region_means=transactions.groupby('Region')['Amount'].mean()
print(region_means)
```

```python
#Modify the 'Amount' for the transaction with TransactionID 102 to 165

transactions.loc[transactions['TransactionID']==102,'Amount']=165
print(transactions)
```

```python
#Add a new column 'Discount' by calculating 10% of the 'Amount'

transactions['Discount']=transactions['Amount']*0.1
print(transactions)
```

```python
#Remove the row with TransactionID 109.

transactions=transactions.drop(transactions[transactions['TransactionID']==109].index)
print(transactions)

```

```python
#Delete the 'Discount' column from the DataFrame.

transactions=transactions.drop('Discount',axis=1)
print(transactions)
```
