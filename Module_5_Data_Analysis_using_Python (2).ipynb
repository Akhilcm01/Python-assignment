DATA CLEANING USING PANDAS

```python
#to see duplicates

df.duplicated().sum()
```

```python
# Remove leading/trailing spaces from all column names

df.columns = df.columns.str.strip()
df.columns

#convert  posted_at column to date format

df["posted_at"]=pd.to_datetime(df["posted_at"],errors="coerce")
print(df.info())



```

```python
#Age cloumn null value imputation and data type changed from float to interger

df["age"].isnull().sum()
df["age"]=df["age"].fillna(df["age"].mean())
df["age"].isnull().sum()
df["age"]=df["age"].astype("int64")
df.info()

```

```python
#Gender cloumn null value imputation

#grouping using value count to understand availoable categories

df["gender"].value_counts(dropna=False)

df["gender"]=df["gender"].fillna("Gender not mentioned")
df.info()
df["gender"].value_counts(dropna=False)
```

```python
#Imputation of na values in 'likes' column using mean

df["likes"]=df["likes"].fillna(df["likes"].mean())
df.info()

#converting data type into int

df["likes"]=df["likes"].astype("int64")
df.info()
```

```python
#Imputation of na values in comments using mean

df["comments"]=df["comments"].fillna(df["comments"].mean())

#converting data type into int
df["comments"]=df["comments"].astype("int64")
df.info()
```

```python
#Imputation of na values in 'shares' column using mean

df["shares"]=df["shares"].fillna(df["shares"].mean())

#converting data type into int

df["shares"]=df["shares"].astype("int64")

df.info()
```

```python
#Check uqique values and categories in sentiment column

df["sentiment"].unique()

df["sentiment"].value_counts(dropna=False)

#Imputation of na values in 'sentiment' column using "Sentiment unknown"


df["sentiment"]=df["sentiment"].fillna("Sentiment unknown")
df.info()
```

```python
#merge

engagement=df[["user_id","likes","comments","shares"]]
print(engagement)

users=df[["user_id","age","gender"]]

merged=pd.merge(engagement,users,on="user_id")
print(merged)
```

FINAL INSIGHTS

```python

#Which post types have highest engagement?


df['engagement_score'] = df['likes'] + df['comments'] + df['shares']
post_type_engagement = df.groupby('post_type')['engagement_score'].mean()
print(post_type_engagement)


#Best-performing content category

category_engagement = df.groupby('post_category')['engagement_score'].mean()
print(category_engagement)


#Which countries have highest avg engagement rate?

country_engagement = df.groupby('country')['engagement_rate'].mean()
print(country_engagement.sort_values(ascending=False))

#How age affects engagement

age_engagement = df.groupby('age')['engagement_score'].mean()
print(age_engagement)

#Performance difference for verified accounts

verified_engagement = df.groupby('is_verified')['engagement_score'].mean()
print(verified_engagement)



```

```python
#Best time of day for impressions


# Extract hour from timestamp
df['hour'] = pd.to_datetime(df['posted_at']).dt.hour

# Average impressions by hour
impressions_by_hour = df.groupby('hour')['impression_count'].mean()
print(impressions_by_hour)


#Device type impact on watch time


device_watch = df.groupby('device_type')['watch_time_sec'].mean()
print(device_watch)


# Which sentiment performs best

sentiment_perf = df.groupby('sentiment')['engagement_score'].mean()
print(sentiment_perf)

#Behavior of negative/neutral sentiment posts

neg_neu = df[df['sentiment'].isin(['negative','neutral'])] \
             .groupby('sentiment')['engagement_score'].mean()
print(neg_neu)



```
